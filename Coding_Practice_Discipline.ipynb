```python
#You are given a nested list. Write a Python function to flatten it (convert it into a single list)
nested = [[1, 2], [3, 4, 5], [6], [7, 8, 9]]

# Output: [1, 2, 3, 4, 5, 6, 7, 8, 9]

def flattened_list(lst):
  flatten=[]
  for i in lst:
    if type(i) is list:
      flatten.extend(flattened_list(i))
    else:
      flatten.append(i)
  return flatten

flattened_list(nested)
```
